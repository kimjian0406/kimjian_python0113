[문제]

'Employee' 클래스를 선언하세요.

이 클래스는 월급인상률을 의미하는 클래스 변수로 'raise_amount'를 갖고 있으며, 기본값은 1.04입니다. 다음 급여는 기본 급여에 월급인상률을 곱한 값으로 결정합니다.

또한, 각 직원의 이름(name)과 급여(salary)를 속성으로 가지며, 'apply_raise' 메소드를 통해 급여에 'raise_amount'를 적용할 수 있어야 합니다.

'Employee' 클래스의 객체 두 개를 생성하고, 한 객체에만 'raise_amount'를 변경한 후, 두 객체의 급여를 증가시켜보세요.<br>
그리고 각 객체의 새로운 급여를 출력하세요.

```python
class Worker:
    raise_percentage = 1.04  # 기본 급여 인상률

    def __init__(self, name, salary):
        """
        Worker 클래스의 생성자. 이름과 급여를 초기화합니다.
        :param name: 직원의 이름
        :param salary: 직원의 초기 급여
        """
        self.name = name
        self.salary = salary

    def apply_raise(self):
        """
        설정된 인상률에 따라 급여를 갱신합니다.
        """
        self.salary = int(self.salary * self.raise_percentage)


# Worker 객체 생성
worker_1 = Worker("길동", 50000)
worker_2 = Worker("현철", 60000)

# 첫 번째 직원의 급여 인상
Worker.raise_percentage = 1.05
worker_1.apply_raise()

# 두 번째 직원의 급여 인상
Worker.raise_percentage = 1.03
worker_2.apply_raise()

# 결과 출력
print(worker_1.salary)  # 가상직원1의 갱신된 급여 출력
print(worker_2.salary)  # 가상직원2의 갱신된 급여 출력
```
