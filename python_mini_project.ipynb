- transaction_type, amount, balance 속성을 초기화하는 생성자를 구현하세요.

```python
# 파일 경로: banking_system/models/transaction.py

class Transaction:
    def __init__(self, transaction_type: str, amount: int, balance: int) -> None:
        """
        Transaction 클래스의 생성자. 거래 유형, 거래 금액, 거래 후 잔고를 초기화합니다.

        :param transaction_type: 거래 유형 (예: "입금", "출금")
        :param amount: 거래 금액
        :param balance: 거래 후 잔고
        """
        self.transaction_type = transaction_type
        self.amount = amount
        self.balance = balance

    def __str__(self) -> str:
        """
        거래 정보를 문자열로 반환합니다.

        :return: 거래 정보 문자열
        """
        return f"거래유형: {self.transaction_type}, 금액: {self.amount}원, 잔고: {self.balance}원"

    def to_tuple(self) -> tuple:
        """
        거래 정보를 튜플 형태로 반환합니다.

        :return: 거래 유형, 금액, 잔고를 포함한 튜플
        """
        return self.transaction_type, self.amount, self.balance

        # Transaction 객체 생성
transaction = Transaction("입금", 30000, 50000)

# __str__ 메서드 출력
print(transaction)  # 출력: 거래유형: 입금, 금액: 10000원, 잔고: 50000원

# to_tuple 메서드 호출
print(transaction.to_tuple())  # 출력: ('입금', 10000, 50000)

```

##### 문제 2.2: 입금 메서드 구현

- 입금을 위한 deposit 메서드를 구현하세요.
- 금액이 양수인지 확인한 후 잔고를 증가시키고, 거래 내역에 추가합니다.
- 타입 힌팅을 사용하여 메서드의 매개변수와 반환 타입을 지정하세요.
    - 함수 시그니처: deposit(self, amount: int) -> None

```python
class Account:
    def __init__(self) -> None:
        """
        Account 클래스의 생성자. 초기 잔액과 거래 기록을 초기화합니다.
        """
        self.__balance = 0  # 초기 잔액은 0으로 설정
        self.transactions = []  # 거래 기록을 저장할 리스트 초기화

    def deposit(self, amount: int) -> None:
        """
        입금을 처리하는 메서드. 금액이 양수인지 확인 후 잔액을 증가시키고 거래 내역을 추가합니다.

        :param amount: 입금 금액
        """
        if amount <= 0:
            print("입금 금액은 양수여야 합니다.")
            return

        self.__balance += amount  # 잔액 증가
        # 거래 내역에 추가
        self.transactions.append(Transaction("입금", amount, self.__balance))
        print(f"{amount}원이 입금되었습니다. 현재 잔액: {self.__balance}원")
```

##### 문제 2.3: 출금 메서드 구현

- 출금을 위한 withdraw 메서드를 구현하세요.
- 금액이 잔고보다 크지 않고 양수인지 확인한 후 잔고를 감소시키고, 거래 내역에 추가합니다.
- 타입 힌팅을 사용하여 메서드의 매개변수와 반환 타입을 지정하세요.
    - 함수 시그니처: withdraw(self, amount: int) -> None

```python
class Account:
    def __init__(self) -> None:
        """
        Account 클래스의 생성자. 초기 잔액과 거래 기록을 초기화합니다.
        """
        self.__balance = 0  # 초기 잔액은 0으로 설정
        self.transactions = []  # 거래 기록을 저장할 리스트 초기화

    def deposit(self, amount: int) -> None:
        """
        입금을 처리하는 메서드. 금액이 양수인지 확인 후 잔액을 증가시키고 거래 내역을 추가합니다.

        :param amount: 입금 금액
        """
        if amount <= 0:
            print("입금 금액은 양수여야 합니다.")
            return

        self.__balance += amount  # 잔액 증가
        # 거래 내역에 추가
        self.transactions.append(Transaction("입금", amount, self.__balance))
        print(f"{amount}원이 입금되었습니다. 현재 잔액: {self.__balance}원")

    def withdraw(self, amount: int) -> None:
        """
        출금을 처리하는 메서드. 금액이 잔고보다 크지 않고 양수인지 확인한 후 잔고를 감소시키고 거래 내역을 추가합니다.

        :param amount: 출금 금액
        """
        if amount <= 0:
            print("출금 금액은 양수여야 합니다.")
            return

        if self.__balance < amount:
            print("잔액이 부족합니다.")
            return

        self.__balance -= amount  # 잔액 감소
        # 거래 내역에 추가
        self.transactions.append(Transaction("출금", amount, self.__balance))
        print(f"{amount}원이 출금되었습니다. 현재 잔액: {self.__balance}원")
```

    - 함수 시그니처: __init__(self) -> None

```python
from banking_system.models.user import User
from banking_system.models.account import Account

class BankingService:
    def __init__(self) -> None:
        """
        BankingService 클래스의 생성자. 사용자 목록을 초기화합니다.
        """
        self.users = []  # 사용자 목록을 초기화

    def add_user(self, username: str) -> None:
        """
        새로운 사용자를 추가하는 메서드.

        :param username: 사용자의 이름
        """
        user = User(username)  # 새로운 User 객체를 생성
        self.users.append(user)  # 사용자 목록에 추가
        print(f"사용자 {username}가 추가되었습니다.")

    def find_user(self, username: str) -> User:
        """
        사용자를 이름으로 찾아 반환하는 메서드.

        :param username: 찾을 사용자의 이름
        :return: User 객체, 사용자가 없으면 None을 반환
        """
        for user in self.users:
            if user.username == username:
                return user  # 사용자가 있으면 해당 User 객체 반환
        print(f"사용자 {username}를 찾을 수 없습니다.")
        return None  # 사용자가 없으면 None 반환

    def user_menu(self, username: str) -> None:
        """
        사용자가 선택할 수 있는 메뉴를 제공하는 메서드.

        :param username: 메뉴를 제공할 사용자 이름
        """
        user = self.find_user(username)  # 사용자 찾기
        if user is None:
            return  # 사용자가 없으면 메뉴 제공 중단

        while True:
            print("\n===== 사용자 메뉴 =====")
            print("1. 잔액 조회")
            print("2. 입금")
            print("3. 출금")
            print("4. 종료")
            choice = input("메뉴를 선택하세요: ")

            if choice == "1":
                print(f"현재 잔액: {user.account.get_balance()} 원")
            elif choice == "2":
                amount = int(input("입금할 금액을 입력하세요: "))
                user.account.deposit(amount)
            elif choice == "3":
                amount = int(input("출금할 금액을 입력하세요: "))
                user.account.withdraw(amount)
            elif choice == "4":
                print("메뉴를 종료합니다.")
                break
            else:
                print("잘못된 선택입니다. 다시 시도하세요.")
```

##### 문제 4.4: 사용자 메뉴 제공 메서드 구현

- 사용자 메뉴를 제공하는 user_menu 메서드를 구현하세요.
- 사용자를 찾고, 입금, 출금, 잔고 확인, 거래 내역 기능을 제공하는 반복 루프를 구현합니다.
- 타입 힌팅을 사용하여 메서드의 매개변수와 반환 타입을 지정하세요.
    - 함수 시그니처: user_menu(self, username: str) -> None

```python
from banking_system.models.user import User

class BankingService:
    def __init__(self) -> None:
        """
        BankingService 클래스의 생성자. 사용자 목록을 초기화합니다.
        """
        self.users: list[User] = []  # 사용자 목록을 초기화

    def add_user(self, username: str) -> None:
        """
        새로운 사용자를 추가하는 메서드.

        :param username: 사용자의 이름
        """
        user = User(username)  # 새로운 User 객체를 생성
        self.users.append(user)  # 사용자 목록에 추가
        print(f"사용자 {username}가 추가되었습니다.")

    def find_user(self, username: str) -> User:
        """
        사용자 목록에서 이름으로 사용자를 찾고, 없으면 예외를 발생시킵니다.

        :param username: 찾을 사용자의 이름
        :return: User 객체
        :raises ValueError: 사용자가 없을 경우
        """
        for user in self.users:
            if user.username == username:
                return user  # 사용자가 있으면 해당 User 객체를 반환
        raise ValueError(f"사용자 {username}를 찾을 수 없습니다.")  # 사용자가 없으면 예외 발생

    def user_menu(self, username: str) -> None:
        """
        사용자가 선택할 수 있는 메뉴를 제공하는 메서드.

        :param username: 메뉴를 제공할 사용자 이름
        """
        try:
            user = self.find_user(username)  # 사용자를 찾기
        except ValueError as e:
            print(e)
            return  # 사용자가 없으면 종료

        while True:
            print("\n===== 사용자 메뉴 =====")
            print("1. 잔액 조회")
            print("2. 입금")
            print("3. 출금")
            print("4. 거래 내역 조회")
            print("5. 종료")
            choice = input("메뉴를 선택하세요: ")

            if choice == "1":
                print(f"현재 잔액: {user.account.get_balance()} 원")
            elif choice == "2":
                amount = int(input("입금할 금액을 입력하세요: "))
                if amount > 0:
                    user.account.deposit(amount)
                    print(f"{amount}원이 입금되었습니다. 현재 잔액: {user.account.get_balance()} 원")
                else:
                    print("입금 금액은 양수여야 합니다.")
            elif choice == "3":
                amount = int(input("출금할 금액을 입력하세요: "))
                if amount > 0:
                    try:
                        user.account.withdraw(amount)
                        print(f"{amount}원이 출금되었습니다. 현재 잔액: {user.account.get_balance()} 원")
                    except ValueError as e:
                        print(e)
                else:
                    print("출금 금액은 양수여야 합니다.")
            elif choice == "4":
                transactions = user.account.get_transactions()
                if transactions:
                    for transaction in transactions:
                        print(transaction)
                else:
                    print("거래 내역이 없습니다.")
            elif choice == "5":
                print("메뉴를 종료합니다.")
                break
            else:
                print("잘못된 선택입니다. 다시 시도하세요.")
```

```python
from typing import Callable

def validate_transaction(func: Callable) -> Callable:
    """
    거래 금액이 0보다 큰지 확인하는 데코레이터.

    :param func: 호출 가능한 함수 (입금, 출금 메서드 등)
    :return: 데코레이터가 적용된 함수
    """
    def wrapper(amount: int, *args, **kwargs):
        if amount <= 0:
            raise NegativeAmountError("금액은 0보다 커야 합니다.")
        return func(amount, *args, **kwargs)
    return wrapper
```

```python
class NegativeAmountError(Exception):
    """금액이 음수일 때 발생하는 예외 클래스"""
    def __init__(self, message: str) -> None:
        super().__init__(message)
```

```python
from banking_system.utils.decorators import validate_transaction
from banking_system.utils.exceptions import NegativeAmountError

class Account:
    def __init__(self, balance: int) -> None:
        self.balance = balance

    @validate_transaction
    def deposit(self, amount: int) -> None:
        self.balance += amount
        print(f"{amount}원이 입금되었습니다. 현재 잔액: {self.balance} 원")

    @validate_transaction
    def withdraw(self, amount: int) -> None:
        if amount > self.balance:
            raise ValueError("잔액이 부족합니다.")
        self.balance -= amount
        print(f"{amount}원이 출금되었습니다. 현재 잔액: {self.balance} 원")


# 예시: 입금 및 출금 시도
account = Account(1000)

try:
    account.deposit(500)  # 정상 입금
    account.deposit(-100)  # 예외 발생: 음수 금액
except NegativeAmountError as e:
    print(e)

try:
    account.withdraw(2000)  # 예외 발생: 잔액 부족
except ValueError as e:
    print(e)
```

#### 문제 5.2: 사용자 정의 예외 클래스 구현

- 사용자 정의 예외 클래스 InsufficientFundsError, NegativeAmountError, UserNotFoundError를 작성하세요.
- 각 예외 클래스에 타입 힌팅을 사용하여 생성자의 매개변수와 반환 타입을 지정하세요.
    - 함수 시그니처:
        - InsufficientFundsError.__init__(self, balance: int) -> None
        - NegativeAmountError.__init__(self) -> None
        - UserNotFoundError.__init__(self, username: str) -> None

```python
class InsufficientFundsError(Exception):
    """잔고 부족 예외 클래스"""
    def __init__(self, balance: int) -> None:
        """
        잔고가 부족할 때 발생하는 예외.

        :param balance: 현재 잔액
        """
        self.balance = balance
        super().__init__(f"잔고 부족. 현재 잔액: {balance} 원.")


class NegativeAmountError(Exception):
    """금액이 음수일 때 발생하는 예외 클래스"""
    def __init__(self) -> None:
        """
        금액이 음수일 때 발생하는 예외.
        """
        super().__init__("거래 금액은 0보다 커야 합니다.")


class UserNotFoundError(Exception):
    """사용자를 찾을 수 없을 때 발생하는 예외 클래스"""
    def __init__(self, username: str) -> None:
        """
        사용자가 존재하지 않을 때 발생하는 예외.

        :param username: 찾을 수 없는 사용자 이름
        """
        self.username = username
        super().__init__(f"사용자를 찾을 수 없습니다: {username}")
```

```python
def withdraw(account_balance: int, amount: int) -> None:
    if amount <= 0:
        raise NegativeAmountError()
    if amount > account_balance:
        raise InsufficientFundsError(account_balance)
    account_balance -= amount
    print(f"출금 완료! 잔액: {account_balance} 원")

def find_user(users: list, username: str) -> str:
    if username not in users:
        raise UserNotFoundError(username)
    return f"사용자 {username}을(를) 찾았습니다."

# 예시: 예외 처리
users = ["Alice", "Bob", "Charlie"]

try:
    withdraw(1000, 1500)  # 잔고 부족 예외 발생
except InsufficientFundsError as e:
    print(e)

try:
    withdraw(1000, -500)  # 음수 금액 예외 발생
except NegativeAmountError as e:
    print(e)

try:
    find_user(users, "David")  # 사용자 찾기 예외 발생
except UserNotFoundError as e:
    print(e)
```
