[문제]

'Rectangle' 클래스를 작성하세요.

이 클래스는 높이(height)와 너비(width)를 속성으로 가지며, 생성자(__init__)를 통해 이 두 속성을 초기화합니다.

'area' 메소드를 추가하여 사각형의 면적을 계산하고 반환하도록 하세요. 높이가 10, 너비가 5인 사각형 객체를 생성하고,<br>
이 객체를 사용하여 면적을 계산해 출력하세요.

```python
class Rectangle:
    def __init__(self, height, width):
        """
        Rectangle 클래스의 생성자. 높이와 너비를 초기화합니다.
        :param height: 사각형의 높이
        :param width: 사각형의 너비
        """
        self.height = height
        self.width = width

    def area(self):
        """
        사각형의 면적을 계산합니다.
        :return: 높이 * 너비
        """
        return self.height * self.width


# 높이 10, 너비 5인 사각형 객체 생성
rect = Rectangle(10, 5)

# 사각형의 면적 계산 및 출력
print("사각형의 면적:", rect.area())
```
